```python
#Task 1

#Input variabel kecepatan dalam km/h
speed_kmh = float(input("Enter speed [km/h]: "))

#Proses pembagian km/h untuk konversi ke m/s
speed_ms = speed_kmh / 3.6

#Output dari hasil input dan konversi kecepatan
print(f"Speed = {speed_ms:.2f} m/s")
```

```python
#Task 2
power_kw = float(input("Masukkan daya [kW]: "))

power = power_kw * 1000

print(f"Power = {power:.2f} W")
```

```python
#Task 3
voltage = float(input("Daya tegangan [V]: "))
current = float(input("Daya arus [A]: "))
operating_time = float(input("Waktu operasi: "))

Power = voltage * current
Energy = Power * operating_time

print(f"===== ELECTRICAL CALCULATION =====")
print(f"Voltage = {voltage:.2f} V")
print(f"Current = {current:.2f} A")
print(f"Power = {Power:.2f} W")
print(f"Operating Time = {operating_time:.2f} hour")
print(f"Energy = {Energy:.2f} Wh")
```
