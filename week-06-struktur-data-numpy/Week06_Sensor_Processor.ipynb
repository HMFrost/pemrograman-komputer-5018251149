```python
import numpy as np

sensor_temp = {
    "id": "KAPAL TEMP",
    "unit": "Celcius",
    "readings": [75.5, 78.2, 82.0, 95.5, 80.1, 79.4, 102.3, 81.0, 77.6, 79.0]
}

def hitung_statistik(data):
    return sum(data) / len(data), max(data)

avg_temp, max_temp = hitung_statistik(sensor_temp["readings"])

temp_array = np.array(sensor_temp["readings"])
THRESHOLD_TEMP = 90.0
anomali_temp = temp_array[temp_array > THRESHOLD_TEMP]

print(f"ID         : {sensor_temp['id']}")
print(f"Rata-rata  : {avg_temp:.2f} {sensor_temp['unit']}")
print(f"Maksimum   : {max_temp} {sensor_temp['unit']}")
print(f"Anomali (> {THRESHOLD_TEMP}°C): {anomali_temp}")
```

```python
import numpy as np
sensor_temp_1 = [75.5, 78.2, 82.0, 95.5, 80.1, 79.4, 102.3, 81.0, 77.6, 79.0]
sensor_temp_2 = [76.0, 77.5, 83.1, 94.0, 79.8, 80.5, 100.1, 82.3, 76.8, 78.5]

matriks_temp = np.array([sensor_temp_1, sensor_temp_2])

print("=== MATRIKS DUA SENSOR TEMPERATUR ===")
print("Bentuk Matriks (Shape):", matriks_temp.shape)
print("\nMatriks Data:\n", matriks_temp)

rata_per_sensor = np.mean(matriks_temp, axis=1)
rata_per_waktu = np.mean(matriks_temp, axis=0)

print(f"Rata-rata Sensor 1 & Sensor 2 (axis=1) : {rata_per_sensor}")
print(f"Rata-rata tiap waktu pembacaan (axis=0): {rata_per_waktu}")
```

1. Axis 0 (Sumbu Baris / Vertikal):
   - Bergerak dari atas ke bawah (menembus baris).
   - Membedakan antar individu sensor (Baris index 0 = Sensor 1, Baris index 1 = Sensor 2).
   - Jika dilakukan 'np.mean(matriks, axis=0)', NumPy menghitung rata-rata nilai dari kedua sensor pada titik waktu/pembacaan yang sama.

2. Axis 1 (Sumbu Kolom / Horizontal):
   - Bergerak dari kiri ke kanan (menembus kolom).
   - Membedakan urutan waktu pembacaan/sampel data (Kolom index 0 = Waktu ke-1, Kolom index 1 = Waktu ke-2, dst).
   - Jika dilakukan 'np.mean(matriks, axis=1)', NumPy menghitung rata-rata keseluruhan waktu untuk masing-masing sensor secara individu.

interpretasi hasil sensor:
1. Berdasarkan hasil pemantauan, sistem berhasil merekam data multivariabel sensor secara terstruktur menggunakan dictionary dan array.
2. Sensor temperatur menunjukkan adanya anomali kenaikan suhu drastis yang melewati batas aman threshold yang telah ditetapkan.
3. Penggunaan library NumPy terbukti sangat efisien dalam melakukan operasi statistik cepat seperti pencarian nilai rata-rata, maksimum, dan minimum.
4. Pemfilteran berbasis kondisi boolean pada array NumPy memudahkan identifikasi cepat terhadap pembacaan sensor yang bermasalah atau melampaui batas.
5. Penyusunan data multi-sensor ke dalam bentuk matriks memudahkan analisis perbandingan kinerja antar sensor berdasarkan sumbu baris maupun kolom waktu.
